```python
import cv2
import numpy as np
from matplotlib import pyplot as plt

# Define our imshow function 
def imshow(title = "Image", image = None, size = 10):
    w, h = image.shape[0], image.shape[1]
    aspect_ratio = w/h
    plt.figure(figsize=(size * aspect_ratio,size))
    plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
    plt.title(title)
    plt.show()
```

# **Watershed Algorthim Theory**

Any grayscale image can be thought of as a topographic surface, with valleys represented by low intensity and peaks and hills by high intensity. 

Using that analogy, this algorithm begins to fill in the troughs, or low areas, with a different colored label (our water). 

Depending on the surrounding peaks and gradients, as the water rises, it will naturally begin to blend the colors of the many valleys. You erect barriers where water mixes in order to prevent that. Once all of the peaks are submerged, you keep filling them with water and erecting obstacles. 

The consequence of segmentation is what you created in the barriers. The "philosophy" underlying the watershed is this. 


In the [CMM webpage](http://cmm.ensmp.fr/~beucher/wtshed.html) on watershed to understand it with the help of some animations.

However, because of noise or other anomalies in the image, their approach results in an oversegmented image. 

As a result, OpenCV developed a marker-based watershed method that lets you choose which valley points should be combined and which ones shouldn't. It assigns many labels to the familiar objects we know. The area that we are certain is the foreground or object should be labeled with one color (or intensity), the area that we are certain is the background or non-object should be labeled with a different color, and the area where we are unsure of anything should be labeled with 0. Our marker is that. Next, utilize the watershed algorithm. Then our marker will be updated with the labels we gave, and the boundaries of objects will have a value of -1.

> **Why `coins.jpg` instead of `liberty.jpeg`?** Watershed segmentation is specifically about splitting *touching/overlapping* objects — it needs a scene where objects actually abut each other, which a single statue portrait doesn't give you. `liberty.jpeg` is still used earlier in this series (arithmetic/bitwise ops), so rather than reuse a poor fit here, this lesson uses scikit-image's classic overlapping-coins sample instead — the same image most watershed tutorials use for exactly this reason.

```python
# Load image
img = cv2.imread('./assets/images/coins.jpg')
imshow("Original image", img)

# Grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Threshold using OTSU
ret, thresh = cv2.threshold(gray, 0, 255, cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)

imshow("Thresholded", thresh)
```

## **Removing the touching masks**

**What `distanceTransform` is doing, concretely:** for every foreground (white) pixel, it computes the straight-line distance to the *nearest* background (black) pixel. A pixel deep in the middle of a coin gets a large value; a pixel near the coin's edge gets a small one. That means the distance map's local peaks sit right at each coin's center — exactly the "sure foreground" seeds watershed needs. Thresholding at `0.7 * dist_transform.max()` keeps only those confident interior peaks, throwing away the ambiguous edge pixels (which is also where two touching coins would otherwise get fused into one blob by a naive threshold). The `unknown` region — foreground minus this confident core — is exactly the boundary strip watershed is left to figure out on its own, flooding outward from each labeled seed until two floods meet and a `-1` boundary line gets drawn between them.

```python
# noise removal
kernel = np.ones((3,3), np.uint8)
opening = cv2.morphologyEx(thresh, cv2.MORPH_OPEN,kernel, iterations = 2)

# sure background area
sure_bg = cv2.dilate(opening, kernel, iterations=3)

# Finding sure foreground area
dist_transform = cv2.distanceTransform(opening, cv2.DIST_L2,5)
ret, sure_fg = cv2.threshold(dist_transform, 0.7 * dist_transform.max(), 255, 0)

# Finding unknown region
sure_fg = np.uint8(sure_fg)
unknown = cv2.subtract(sure_bg, sure_fg)

imshow("SureFG", sure_fg)
imshow("SureBG", sure_bg)
imshow("unknown", unknown)
```

```python
# Marker labelling
# Connected Components determines the connectivity of blob-like regions in a binary image.
ret, markers = cv2.connectedComponents(sure_fg)

# Add one to all labels so that sure background is not 0, but 1
markers = markers+1

# Now, mark the region of unknown with zero
markers[unknown==255] = 0

markers = cv2.watershed(img,markers)
img[markers == -1] = [255,0,0]

imshow("img", img)
```

### Worth trying next

- Count the coins: `ret, markers = cv2.connectedComponents(sure_fg)` before running watershed already gives you a rough per-coin count via `markers.max()` — compare it to the post-watershed boundary count.
- Try skipping the distance-transform step and watershed directly on `opening` — you'll see why the sure-foreground step matters (over-segmentation).
