```python
# Load an image using 'imread' specifying the path to image
image = cv2.imread('./assets/images/castara.jpeg')
```

```python
from matplotlib import pyplot as plt

#Show the image with matplotlib
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.show()
```

```python
def imshow(title = "", image = None):
    plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
    plt.title(title)
    plt.show()
```

```python
# Let's test it out
imshow("Displaying Our First Image", image)
```

```python
# Simply use 'imwrite' specificing the file name and the image to be saved
cv2.imwrite('output.jpg', image)
```

```python
# Or save it as a PNG (Portable Network Graphics) which is a Lossless bitmap image format
cv2.imwrite('output.png', image)
```

```python
# Import numpy
import numpy as np

print(image.shape) # (height, width, 3 for color images(RGB => 3 Channels)))
```

```python
# To access a dimension, simply index it by using 0, 1 or 2.
print('Height: {} pixels'. format(int((image.shape[0])))) 
print('Width: {} pixels'. format(int(image.shape[1])))
print('Depth: {}'. format(int(image.shape[2]))) # color images have 3 channels (RGB) ; black&white images have only one channel 
```
